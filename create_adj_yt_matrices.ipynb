### Create yt dataset


```python
import networkx as nx
import pickle
import os
from tqdm import tqdm
```

```python
G = nx.Graph()
path = './dataset/youtube/'

G = nx.Graph()
with open(path + "com-Youtube.ungraph.txt", "rt") as f:
    for line in f:
        if not line.startswith("#"):
            u, v = map(int, line.split())
            G.add_edge(u, v)

communities = []
with open(path + "com-Youtube.all.cmty.txt", "rt") as f:
    for line in f:
        nodes = list(map(int, line.split()))
        communities.append(nodes)
```

```python
for index in tqdm(range(len(communities))):
    nodes = communities[index]
    if len(nodes) <= 33:
        continue
    H = G.subgraph(nodes)
    A_H = nx.adjacency_matrix(H).toarray()
    file_name = f"graph_{index}.pkl"
    with open(path + "pickle/" + file_name, "wb") as f:
        pickle.dump([A_H], f)
    os.system(f"python3.10 SIGL/Main.py --n_graphon 1 --Res 32 --Adjs_dir ./dataset/youtube/pickle/{file_name} --test_name {index} > /dev/null")
```
