```python
import torch
from diffusers import DiffusionPipeline

model_id = "mikymatt/ddpm-ema-flowers-64"   # es. "mioHubUser/ddpm‑ema‑flowers‑64"

# 1) carica la pipeline dal Hub
pipe = DiffusionPipeline.from_pretrained(model_id)

# 2) sposta il modello su device
device = "cuda" if torch.cuda.is_available() else "cpu"
pipe = pipe.to(device)

# 3) esegui inferenza
output = pipe()  # oppure passa gli argomenti necessari
images = output.images

# salva la prima
images[0].save("inferenza_output.png")
```
